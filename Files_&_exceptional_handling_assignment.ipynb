Theory Questions

**1. What is the difference between interpreted and compiled languages?**
- Interpreted languages, like Python, run directly from the source code line by line using an interpreter. Compiled languages, like C++, are first converted into machine code by a compiler before execution. The key difference is how they execute: interpreters are slower but easier to debug, while compilers are faster but take a bit more setup.

**2. What is exception handling in Python?**
- Exception handling is Python's way of dealing with unexpected errors during code execution. Instead of crashing, we can use `try-except` blocks to gracefully manage what happens when something goes wrong.

**3. What is the purpose of the finally block in exception handling?**
- The `finally` block always runs, no matter what. It's perfect for cleaning up, like closing a file or releasing a resource, whether an error occurred or not.

**4. What is logging in Python?**
- Logging is like smart printing. Instead of `print()`, we log messages with levels like INFO, ERROR, or DEBUG. These logs can be stored in files and are super helpful when debugging or tracking the behavior of an app.

**5. What is the significance of the __del__ method in Python?**
- The `__del__` method is a destructor. It’s called when an object is about to be destroyed. If you need to clean up something like closing files or connections, `__del__` is a good place to do it.

**6. What is the difference between import and from ... import in Python?**
- Using `import math` means you’ll write `math.sqrt(4)`. Using `from math import sqrt` lets you just write `sqrt(4)`. Both work, it’s just about how directly you want to access functions.

**7. How can you handle multiple exceptions in Python?**
- You can catch multiple exceptions using a tuple:
```python
try:
    risky_code()
except (ValueError, TypeError) as e:
    print("Handled error:", e)
```

**8. What is the purpose of the with statement when handling files in Python?**
- `with` is used to handle files safely. It automatically closes the file when you're done—even if there’s an error.
```python
with open("data.txt") as f:
    contents = f.read()
```

**9. What is the difference between multithreading and multiprocessing?**
- Multithreading runs multiple threads in the same memory space. It's best for I/O tasks. Multiprocessing creates separate processes, ideal for CPU-heavy tasks.

**10. What are the advantages of using logging in a program?**
- Logging helps track events in a program. It's flexible, can be turned off or directed to a file, and supports different levels (DEBUG, INFO, etc.).

**11. What is memory management in Python?**
- Python handles memory using automatic allocation and garbage collection. It keeps track of object references and cleans up unused ones.

**12. What are the basic steps involved in exception handling in Python?**
1. Wrap code in `try` block
2. Catch errors with `except`
3. Use `else` if no errors occur
4. Use `finally` to always run cleanup code

**13. Why is memory management important in Python?**
- It ensures your program runs efficiently without leaking memory. Poor memory management can slow things down or crash your program.

**14. What is the role of try and except in exception handling?**
- They allow you to run risky code and handle problems without crashing the program.

**15. How does Python's garbage collection system work?**
- It uses reference counting and a garbage collector that finds and removes objects no longer in use.

**16. What is the purpose of the else block in exception handling?**
- `else` runs only if there was no exception in the `try` block. It's useful for code that should only run when everything goes well.

**17. What are the common logging levels in Python?**
- DEBUG
- INFO
- WARNING
- ERROR
- CRITICAL

**18. What is the difference between os.fork() and multiprocessing in Python?**
- `os.fork()` creates a child process (Unix-only). `multiprocessing` is a cross-platform module that makes parallel processing easier and safer.

**19. What is the importance of closing a file in Python?**
- Closing a file frees system resources and ensures all data is saved correctly.

**20. What is the difference between file.read() and file.readline()?**
- `read()` reads the entire file
- `readline()` reads one line at a time

**21. What is the logging module in Python used for?**
- It's used for recording log messages. It gives more control than print statements and supports levels, formatting, and file writing.

**22. What is the os module in Python used for in file handling?**
- The `os` module lets you interact with the operating system. You can create, rename, delete, and check files or directories.

**23. What are the challenges associated with memory management in Python?**
- Circular references
- High memory usage if not optimized
- Unpredictable cleanup due to garbage collection

**24. How do you raise an exception manually in Python?**
```python
raise ValueError("This is a manual error")
```

**25. Why is it important to use multithreading in certain applications?**
- It helps when you want to do multiple things at once, especially I/O tasks like downloading files or user input, without freezing your app.
